```python
import random

# 1. Número aleatorio entre 1 y 100
numero_secreto = random.randint(1, 100)

# 3. Lista que almacena los intentos válidos
intentos = []

# 2. Introducción y reglas
print("=" * 50)
print("   JUEGO DE ADIVINAR EL NÚMERO")
print("=" * 50)
print("He pensado un número entero entre 1 y 100.")
print("Reglas:")
print(" - Si tu número es menor que 1 o mayor que 100: FUERA DE LÍMITES.")
print(" - En el primer intento: ¡CALIENTE! si estás a 10 o menos, ¡FRIO! si no.")
print(" - En los siguientes: ¡MAS CALIENTE! si te acercas, ¡MAS FRIO! si te alejas.")
print(" - Cuando aciertes, te diré cuántos intentos has necesitado.")
print()

while True:
    entrada = input("Tu número: ")

    # Validar que sea un entero
    try:
        numero = int(entrada)
    except ValueError:
        print("Por favor, escribe un número entero.")
        continue

    # Regla 1: fuera de límites (no cuenta como intento)
    if numero < 1 or numero > 100:
        print("FUERA DE LÍMITES")
        continue

    # Guardamos el intento
    intentos.append(numero)

    # Regla 4: acierto
    if numero == numero_secreto:
        print(f"¡Has acertado! El número era {numero_secreto}.")
        print(f"Lo has conseguido en {len(intentos)} intentos.")
        break

    distancia_actual = abs(numero_secreto - numero)

    if len(intentos) == 1:
        # Regla 2: primer intento
        if distancia_actual <= 10:
            print("¡CALIENTE!")
        else:
            print("¡FRIO!")
    else:
        # Regla 3: comparar con el intento anterior
        distancia_anterior = abs(numero_secreto - intentos[-2])
        if distancia_actual < distancia_anterior:
            print("¡MAS CALIENTE!")
        elif distancia_actual > distancia_anterior:
            print("¡MAS FRIO!")
        else:
            print("Estás igual de lejos que antes.")
    continue
```
